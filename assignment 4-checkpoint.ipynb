````python
from pathlib import Path
import zipfile

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, balanced_accuracy_score, f1_score
from sklearn.model_selection import train_test_split
from sklearn.neighbors import NearestNeighbors
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

DATA_DIR = Path.home() / "Desktop" / "assignment 4"
adult = pd.read_csv(DATA_DIR / "adult" / "adult.csv")
compas = pd.read_csv(DATA_DIR / "compas" / "cox-violent-parsed_filt.csv")
german_columns = [f"A{i}" for i in range(1, 21)] + ["credit_risk"]
german_file = DATA_DIR / "statlog+german+credit+data" / "german.data"
german_zip = DATA_DIR / "statlog+german+credit+data.zip"

if german_file.exists():
    german = pd.read_csv(german_file, sep=r"\s+", header=None, names=german_columns)
elif german_zip.exists():
    with zipfile.ZipFile(german_zip) as archive:
        german = pd.read_csv(archive.open("german.data"), sep=r"\s+", header=None, names=german_columns)
else:
    raise FileNotFoundError("Place german.data in statlog+german+credit+data/ or provide the German ZIP.")

datasets = {"Adult": adult, "German Credit": german, "COMPAS": compas}
for name, df in datasets.items():
    print(f"\n{name}: {df.shape}")
    print("Columns:", df.columns.tolist())
    print("\nData types:\n", df.dtypes)
    missing = df.isna().sum()
    print("\nMissing values:\n", missing[missing > 0])
````

## Define Targets and Protected Attributes

Set each dataset’s target, favorable outcome, and protected attributes. Inspect the resulting group counts before choosing how to compare groups.

````python
target_config = {
    "Adult": {"target": "income", "favorable": ">50K"},
    "German Credit": {"target": "credit_risk", "favorable": 1},
    "COMPAS": {"target": "is_violent_recid", "favorable": 0},
}

adult_groups = adult.assign(
    sex_group=adult["sex"],
    race_group=adult["race"],
    age_group=adult["age"].map(lambda age: "25+" if age >= 25 else "under 25"),
)

german_groups = german.assign(
    sex_group=german["A9"].map(
        lambda value: "Female" if value in ["A92", "A95"] else "Male"
    ),
    age_group=german["A2"].map(lambda age: "25+" if age >= 25 else "under 25"),
)

compas_groups = compas.assign(
    sex_group=compas["sex"],
    race_group=compas["race"],
    age_group=compas["age"].map(
        lambda age: "25+" if age >= 25 else "under 25"
    ),
)

grouped_datasets = {
    "Adult": adult_groups,
    "German Credit": german_groups,
    "COMPAS": compas_groups,
}

protected_columns = {
    "Adult": ["sex_group", "race_group", "age_group"],
    "German Credit": ["sex_group", "age_group"],
    "COMPAS": ["sex_group", "race_group", "age_group"],
}


for dataset_name, df in grouped_datasets.items():
    target = target_config[dataset_name]["target"]

    print(f"\n{dataset_name} — target: {target}")
    for protected_column in protected_columns[dataset_name]:
        print(f"\n{protected_column}")
        print(pd.crosstab(df[protected_column], df[target], margins=True))
````



| Dataset | Target | Favorable outcome | Protected attributes |
|---|---|---|---|
| Adult | `income` | `>50K` | `sex`, `race`, age group |
| German Credit | `credit_risk` | `1` (good credit) | Sex group derived from `A9`, age group derived from `A2` |
| COMPAS | `is_violent_recid` | `0` (no violent recidivism) | `sex`, `race`, age group |

All protected groups contain both target classes. However, some outcomes and groups are imbalanced, particularly violent recidivism in COMPAS and `>50K` income among Adult participants under 25. Fairness results for these groups may be less stable.

## Baseline Fairness of the Original Labels

Calculate DP_Diff and DIR from the original labels before splitting or training. Consistency is calculated once per dataset using five nearest neighbors in encoded feature space. The score is `1 - mean(abs(y_i - mean(neighbor labels)))`. Adult `?` values are treated as missing, matching model preprocessing.


````python
reference_groups = {
    "Adult": {
        "sex_group": "Male",
        "race_group": "White",
        "age_group": "25+",
    },
    "German Credit": {
        "sex_group": "Male",
        "age_group": "25+",
    },
    "COMPAS": {
        "sex_group": "Male",
        "race_group": "Caucasian",
        "age_group": "25+",
    },
}

# Exclude targets, identifiers, and COMPAS fields recorded after assessment.
excluded_features = {
    "Adult": ["income"],
    "German Credit": ["credit_risk"],
    "COMPAS": [
        "is_violent_recid",
        "violent_recid",
        "is_recid",
        "event",
        "r_charge_degree",
        "r_days_from_arrest",
        "r_offense_date",
        "r_charge_desc",
        "r_jail_in",
        "vr_charge_degree",
        "vr_offense_date",
        "vr_charge_desc",
        "id",
        "name",
        "first",
        "last",
        "dob",
        "screening_date",
        "v_screening_date",
        "in_custody",
        "out_custody",
        "decile_score",
        "decile_score.1",
        "score_text",
        "v_decile_score",
        "v_score_text",
        "type_of_assessment",
        "v_type_of_assessment",
        "c_jail_in",
        "c_jail_out",
        "c_days_from_compas",
        "priors_count.1",
    ],
}
baseline_rows = []
consistency_rows = []

for dataset_name, df in grouped_datasets.items():
    target = target_config[dataset_name]["target"]
    favorable = target_config[dataset_name]["favorable"]

    # Encode the favorable original label as 1 and all other labels as 0.
    y = (df[target] == favorable).astype(int).to_numpy()

    # Keep raw protected attributes in X; exclude derived grouping columns.
    drop_columns = excluded_features[dataset_name] + protected_columns[dataset_name]
    X = df.drop(columns=drop_columns, errors="ignore").copy()

    if dataset_name == "Adult":
        X = X.replace("?", np.nan)

    numeric_columns = X.select_dtypes(include=["number", "bool"]).columns
    categorical_columns = X.select_dtypes(
        include=["object", "str", "category"]
    ).columns

    preprocessing = ColumnTransformer(
        transformers=[
            (
                "numeric",
                make_pipeline(
                    SimpleImputer(strategy="median"),
                    StandardScaler(),
                ),
                numeric_columns,
            ),
            (
                "categorical",
                make_pipeline(
                    SimpleImputer(strategy="most_frequent"),
                    OneHotEncoder(handle_unknown="ignore"),
                ),
                categorical_columns,
            ),
        ]
    )

    X_encoded = preprocessing.fit_transform(X)

    # AIF360-style consistency using k=5 nearest neighbors.
    k = 5
    neighbor_model = NearestNeighbors(n_neighbors=k + 1).fit(X_encoded)
    neighbor_indices = neighbor_model.kneighbors(
        X_encoded, return_distance=False
    )

    consistency_neighbors = np.empty((len(df), k), dtype=int)
    for row_index, indices in enumerate(neighbor_indices):
        indices = indices[indices != row_index]
        consistency_neighbors[row_index] = indices[:k]

    consistency = 1 - np.abs(
        y - y[consistency_neighbors].mean(axis=1)
    ).mean()

    consistency_rows.append({
        "dataset": dataset_name,
        "consistency": consistency,
        "k_neighbors": k,
    })

    # DP_Diff and DIR from the original labels, for each protected group.
    for protected_column in protected_columns[dataset_name]:
        reference = reference_groups[dataset_name][protected_column]
        reference_mask = df[protected_column].to_numpy() == reference
        reference_rate = y[reference_mask].mean()

        for group in df[protected_column].dropna().unique():
            group_mask = df[protected_column].to_numpy() == group
            group_rate = y[group_mask].mean()

            baseline_rows.append({
                "dataset": dataset_name,
                "protected_attribute": protected_column,
                "group": group,
                "reference_group": reference,
                "favorable_rate": group_rate,
                "DP_Diff": group_rate - reference_rate,
                "DIR": (
                    group_rate / reference_rate
                    if reference_rate != 0
                    else np.nan
                ),
            })

baseline_fairness = pd.DataFrame(baseline_rows)
baseline_consistency = pd.DataFrame(consistency_rows)

display(baseline_fairness)
display(baseline_consistency)
````

The favorable outcome is encoded as 1. DP_Diff is the group favorable-label rate minus the reference-group rate; DIR is the ratio of those rates. Small groups and imbalanced labels can make estimates unstable.


## Stratified Train/Test Split

Split each dataset into training and test sets while preserving the favorable-outcome class proportions. Keep protected-group values separate for fairness evaluation.

````python
from sklearn.model_selection import train_test_split

splits = {}

for dataset_name, df in grouped_datasets.items():
    target = target_config[dataset_name]["target"]
    favorable = target_config[dataset_name]["favorable"]

    y = (df[target] == favorable).astype(int)

    # Use the same feature exclusions defined for the baseline analysis.
    drop_columns = excluded_features[dataset_name] + protected_columns[dataset_name]
    X = df.drop(columns=drop_columns, errors="ignore")
    groups = df[protected_columns[dataset_name]]

    X_train, X_test, y_train, y_test, groups_train, groups_test = train_test_split(
        X,
        y,
        groups,
        test_size=0.20,
        random_state=42,
        stratify=y,
    )

    splits[dataset_name] = {
        "X_train": X_train,
        "X_test": X_test,
        "y_train": y_train,
        "y_test": y_test,
        "groups_train": groups_train,
        "groups_test": groups_test,
    }

    print(
        f"{dataset_name}: "
        f"train={len(X_train)}, test={len(X_test)}, "
        f"train favorable rate={y_train.mean():.3f}, "
        f"test favorable rate={y_test.mean():.3f}"
    )
````

## Preprocess Features Using Training Data

Fit imputation, scaling, and categorical encoding on each training set, then apply the fitted transformations to its test set. Raw protected attributes are retained as model features; derived group columns are kept separately for Reweighing and fairness evaluation. Preprocessing is fit on training data only.


````python
processed_splits = {}
preprocessors = {}

for dataset_name, split in splits.items():
    X_train = split["X_train"].copy()
    X_test = split["X_test"].copy()

    # The Adult CSV uses '?' to represent missing categorical values.
    if dataset_name == "Adult":
        X_train = X_train.replace("?", np.nan)
        X_test = X_test.replace("?", np.nan)

    numeric_columns = X_train.select_dtypes(
        include=["number", "bool"]
    ).columns.tolist()
    categorical_columns = X_train.select_dtypes(
        include=["object", "str", "category"]
    ).columns.tolist()

    preprocessor = ColumnTransformer(
        transformers=[
            (
                "numeric",
                make_pipeline(
                    SimpleImputer(strategy="median"),
                    StandardScaler(),
                ),
                numeric_columns,
            ),
            (
                "categorical",
                make_pipeline(
                    SimpleImputer(strategy="most_frequent"),
                    OneHotEncoder(handle_unknown="ignore"),
                ),
                categorical_columns,
            ),
        ]
    )

    X_train_processed = preprocessor.fit_transform(X_train)
    X_test_processed = preprocessor.transform(X_test)

    preprocessors[dataset_name] = preprocessor
    processed_splits[dataset_name] = {
        **split,
        "X_train_processed": X_train_processed,
        "X_test_processed": X_test_processed,
    }

    print(
        f"{dataset_name}: "
        f"train features={X_train_processed.shape}, "
        f"test features={X_test_processed.shape}"
    )
````

## Train Baseline Classifiers

Train logistic regression models on the original training data. Evaluate predictive performance on the held-out test data.

````python
baseline_models = {}
baseline_predictions = {}
performance_rows = []

for dataset_name, split in processed_splits.items():
    model = LogisticRegression(max_iter=2000)
    model.fit(
        split["X_train_processed"],
        split["y_train"],
    )
    y_pred = model.predict(split["X_test_processed"])

    baseline_models[dataset_name] = model
    baseline_predictions[dataset_name] = y_pred

    y_true = split["y_test"]
    performance_rows.append({
        "dataset": dataset_name,
        "accuracy": accuracy_score(y_true, y_pred),
        "balanced_accuracy": balanced_accuracy_score(y_true, y_pred),
        "macro_f1": f1_score(y_true, y_pred, average="macro"),
        "unfavorable_class_f1": f1_score(
            y_true, y_pred, pos_label=0, zero_division=0
        ),
    })

baseline_performance = pd.DataFrame(performance_rows)
baseline_performance
````

## Apply Reweighing to Training Data

For each protected attribute, assign each training observation the weight `w(a, y) = P(A=a) * P(Y=y) / P(A=a, Y=y)`. This applies the Reweighing cell-weight formula to categories of the attribute. Weights are calculated from training data only, normalized to mean 1, and supplied to Logistic Regression as `sample_weight`. Test data is never used to calculate weights.


````python
def calculate_reweighing_weights(protected_values, labels):
    data = pd.DataFrame({
        "group": np.asarray(protected_values),
        "label": np.asarray(labels),
    })

    group_probability = data["group"].value_counts(normalize=True)
    label_probability = data["label"].value_counts(normalize=True)
    joint_probability = data.groupby(["group", "label"]).size() / len(data)

    weights = np.array([
        group_probability[group] * label_probability[label]
        / joint_probability.loc[(group, label)]
        for group, label in zip(data["group"], data["label"])
    ])

    return weights / weights.mean()


reweighed_models = {}
reweighed_predictions = {}
reweighing_weight_summary = []

for dataset_name, split in processed_splits.items():
    for protected_column in protected_columns[dataset_name]:
        weights = calculate_reweighing_weights(
            split["groups_train"][protected_column],
            split["y_train"],
        )

        model = LogisticRegression(max_iter=2000)
        model.fit(
            split["X_train_processed"],
            split["y_train"],
            sample_weight=weights,
        )

        y_pred = model.predict(split["X_test_processed"])
        model_key = (dataset_name, protected_column)

        reweighed_models[model_key] = model
        reweighed_predictions[model_key] = y_pred

        reweighing_weight_summary.append({
            "dataset": dataset_name,
            "protected_attribute": protected_column,
            "minimum_weight": weights.min(),
            "maximum_weight": weights.max(),
            "mean_weight": weights.mean(),
        })

reweighing_weight_summary = pd.DataFrame(reweighing_weight_summary)
reweighing_weight_summary
````

## Evaluate Baseline and Reweighed Models on the Test Set

Compare predictions on the held-out test data. Report predictive performance, DP_Diff, DIR, and Consistency for the baseline model and each Reweighing run.

````python
def calculate_consistency(X, labels, k=5):
    labels = np.asarray(labels)
    k = min(k, len(labels) - 1)

    neighbor_model = NearestNeighbors(n_neighbors=k + 1).fit(X)
    neighbor_indices = neighbor_model.kneighbors(
        X, return_distance=False
    )

    neighbors = np.empty((len(labels), k), dtype=int)
    for row_index, indices in enumerate(neighbor_indices):
        indices = indices[indices != row_index]
        neighbors[row_index] = indices[:k]

    return 1 - np.abs(
        labels - labels[neighbors].mean(axis=1)
    ).mean()


performance_rows = []
fairness_rows = []
consistency_rows = []

for dataset_name, split in processed_splits.items():
    model_runs = [
        ("Baseline", None, baseline_predictions[dataset_name])
    ]

    for protected_column in protected_columns[dataset_name]:
        key = (dataset_name, protected_column)
        model_runs.append((
            "Reweighed",
            protected_column,
            reweighed_predictions[key],
        ))

    for model_name, reweighed_attribute, y_pred in model_runs:
        y_true = split["y_test"].to_numpy()
        y_pred = np.asarray(y_pred)

        performance_rows.append({
            "dataset": dataset_name,
            "model": model_name,
            "reweighed_attribute": reweighed_attribute or "N/A",
            "accuracy": accuracy_score(y_true, y_pred),
            "balanced_accuracy": balanced_accuracy_score(y_true, y_pred),
            "macro_f1": f1_score(y_true, y_pred, average="macro"),
            "unfavorable_class_f1": f1_score(
                y_true, y_pred, pos_label=0, zero_division=0
            ),
        })

        consistency_rows.append({
            "dataset": dataset_name,
            "model": model_name,
            "reweighed_attribute": reweighed_attribute or "N/A",
            "consistency": calculate_consistency(
                split["X_test_processed"], y_pred
            ),
            "k_neighbors": min(5, len(y_pred) - 1),
        })

        attributes_to_evaluate = (
            protected_columns[dataset_name]
            if reweighed_attribute is None
            else [reweighed_attribute]
        )

        for protected_column in attributes_to_evaluate:
            group_values = split["groups_test"][protected_column].to_numpy()
            reference = reference_groups[dataset_name][protected_column]
            reference_mask = group_values == reference

            if not reference_mask.any():
                reference_rate = np.nan
            else:
                reference_rate = y_pred[reference_mask].mean()

            for group in pd.unique(group_values):
                group_mask = group_values == group
                group_rate = y_pred[group_mask].mean()

                fairness_rows.append({
                    "dataset": dataset_name,
                    "model": model_name,
                    "reweighed_attribute": reweighed_attribute or "N/A",
                    "protected_attribute": protected_column,
                    "group": group,
                    "reference_group": reference,
                    "favorable_rate": group_rate,
                    "DP_Diff": group_rate - reference_rate,
                    "DIR": (
                        group_rate / reference_rate
                        if pd.notna(reference_rate) and reference_rate != 0
                        else np.nan
                    ),
                })

test_performance = pd.DataFrame(performance_rows)
test_fairness = pd.DataFrame(fairness_rows)
test_consistency = pd.DataFrame(consistency_rows)

display(test_performance)
display(test_fairness)
display(test_consistency)
````

## Compare Baseline and Reweighed Results

Compare baseline and Reweighed model predictions on the same held-out observations. Each Reweighed run is evaluated for the protected attribute used to calculate its weights. Negative absolute-gap changes indicate movement toward parity; interpret fairness measures alongside predictive performance.


````python
# Compare test-set fairness by dataset, protected attribute, and group.
fairness_baseline = (
    test_fairness[test_fairness["model"] == "Baseline"]
    [["dataset", "protected_attribute", "group", "DP_Diff", "DIR"]]
    .rename(columns={
        "DP_Diff": "DP_Diff_baseline",
        "DIR": "DIR_baseline",
    })
)

fairness_reweighed = (
    test_fairness[test_fairness["model"] == "Reweighed"]
    [["dataset", "protected_attribute", "group", "DP_Diff", "DIR"]]
    .rename(columns={
        "DP_Diff": "DP_Diff_reweighed",
        "DIR": "DIR_reweighed",
    })
)

fairness_comparison = fairness_baseline.merge(
    fairness_reweighed,
    on=["dataset", "protected_attribute", "group"],
)

fairness_comparison["absolute_DP_gap_change"] = (
    fairness_comparison["DP_Diff_reweighed"].abs()
    - fairness_comparison["DP_Diff_baseline"].abs()
)

fairness_comparison["DIR_distance_from_1_change"] = (
    (fairness_comparison["DIR_reweighed"] - 1).abs()
    - (fairness_comparison["DIR_baseline"] - 1).abs()
)


# Compare predictive performance.
performance_baseline = (
    test_performance[test_performance["model"] == "Baseline"]
    .drop(columns=["model", "reweighed_attribute"])
)

performance_reweighed = (
    test_performance[test_performance["model"] == "Reweighed"]
    .rename(columns={"reweighed_attribute": "protected_attribute"})
    .drop(columns=["model"])
)

performance_comparison = performance_reweighed.merge(
    performance_baseline,
    on="dataset",
    suffixes=("_reweighed", "_baseline"),
)

for metric in [
    "accuracy",
    "balanced_accuracy",
    "macro_f1",
    "unfavorable_class_f1",
]:
    performance_comparison[f"{metric}_change"] = (
        performance_comparison[f"{metric}_reweighed"]
        - performance_comparison[f"{metric}_baseline"]
    )


# Compare prediction Consistency.
consistency_baseline = (
    test_consistency[test_consistency["model"] == "Baseline"]
    [["dataset", "consistency"]]
    .rename(columns={"consistency": "consistency_baseline"})
)

consistency_reweighed = (
    test_consistency[test_consistency["model"] == "Reweighed"]
    .rename(columns={"reweighed_attribute": "protected_attribute"})
    [["dataset", "protected_attribute", "consistency"]]
    .rename(columns={"consistency": "consistency_reweighed"})
)

consistency_comparison = consistency_reweighed.merge(
    consistency_baseline,
    on="dataset",
)

consistency_comparison["consistency_change"] = (
    consistency_comparison["consistency_reweighed"]
    - consistency_comparison["consistency_baseline"]
)

display(fairness_comparison)
display(performance_comparison)
display(consistency_comparison)
````

## Visualize Test-Set DP Gaps

The plots compare absolute DP_Diff for each protected-group comparison before and after Reweighing. Smaller absolute gaps are closer to parity, but do not establish fairness on their own.


````python
plot_data = fairness_comparison.copy()
reference_groups_for_plot = {
    "Adult": {"sex_group": "Male", "race_group": "White", "age_group": "25+"},
    "German Credit": {"sex_group": "Male", "age_group": "25+"},
    "COMPAS": {"sex_group": "Male", "race_group": "Caucasian", "age_group": "25+"},
}
plot_data = plot_data[
    plot_data.apply(lambda row: row["group"] != reference_groups_for_plot[row["dataset"]][row["protected_attribute"]], axis=1)
].copy()
plot_data["comparison"] = plot_data["dataset"] + " | " + plot_data["protected_attribute"] + " | " + plot_data["group"].astype(str)

fig, axes = plt.subplots(3, 1, figsize=(11, 12))
for ax, (dataset_name, df) in zip(axes, plot_data.groupby("dataset", sort=False)):
    df = df.iloc[::-1]
    y = np.arange(len(df))
    ax.barh(y - 0.18, df["DP_Diff_baseline"].abs(), height=0.34, label="Baseline")
    ax.barh(y + 0.18, df["DP_Diff_reweighed"].abs(), height=0.34, label="Reweighed")
    ax.set_yticks(y, df["comparison"])
    ax.set_title(dataset_name)
    ax.set_xlabel("Absolute DP_Diff")
    ax.grid(axis="x", alpha=0.25)
axes[0].legend()
fig.suptitle("Held-out Test Set: Absolute DP_Diff Before and After Reweighing")
fig.tight_layout()
plt.show()
````

````python

from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

compas_split = processed_splits["COMPAS"]
y_true = compas_split["y_test"]

compas_models = {
    "Baseline": baseline_predictions["COMPAS"],
    "Reweighed - Sex": reweighed_predictions[
        ("COMPAS", "sex_group")
    ],
    "Reweighed - Race": reweighed_predictions[
        ("COMPAS", "race_group")
    ],
    "Reweighed - Age": reweighed_predictions[
        ("COMPAS", "age_group")
    ],
}

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

for ax, (model_name, y_pred) in zip(
    axes.flatten(), compas_models.items()
):
    cm = confusion_matrix(y_true, y_pred, labels=[0, 1])

    display = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=["Unfavourable (0)", "Favourable (1)"]
    )

    display.plot(ax=ax, cmap="Blues", values_format="d")
    ax.set_title(model_name)

plt.tight_layout()
plt.show()
````

## Results and Discussion

Reweighing narrowed several measured DP_Diff gaps for Adult and German Credit, but effects differed by protected attribute. Adult’s under-25 income gap remained substantial, and the Asian-Pac-Islander gap increased slightly. German Credit’s sex and age gaps narrowed modestly, while balanced accuracy improved slightly. The German Credit test set contains 200 records, and several race groups are small, so estimates may be unstable.

For COMPAS, the target is `is_violent_recid` and the favorable outcome is 0 (no recorded violent recidivism). Reweighing made little difference. Balanced accuracy stayed near 0.50, and F1 for violent recidivism remained near zero. Its near-parity DP_Diff and DIR therefore reflect predictions that are almost always favorable, not evidence of useful or equitable prediction. The near-1 test Consistency likewise reflects nearly constant predictions.

Original-label Consistency is calculated on each full dataset before splitting; model Consistency is calculated on held-out test predictions. Compare baseline and Reweighed model Consistency on the same test set. Results are conditional on the chosen group definitions, features, and classifier. Raw protected attributes are retained as model features.
